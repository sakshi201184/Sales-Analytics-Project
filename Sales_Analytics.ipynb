# Sales Analytics Dashboard Using Python and Power BI

## Week 6 Final Data Analytics Project
This notebook performs data preparation, exploratory data analysis, visualization, and business insights.

### 1. Import Libraries
```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### 2. Load Dataset
```python
df = pd.read_csv("Sales_Analytics_Week6_Dataset.csv")
df.head()
```

### 3. Data Inspection
```python
df.shape
df.info()
df.describe()
df.isnull().sum()
```

### 4. Data Cleaning
```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")
df["Payment_Mode"] = df["Payment_Mode"].fillna("Unknown")
df["Customer_Type"] = df["Customer_Type"].fillna("Unknown")
df = df.drop_duplicates()
df = df[(df["Quantity"] > 0) & (df["Sales"] >= 0) & (df["Profit"] >= 0)]
```

### 5. Feature Creation
```python
df["Month"] = df["Order_Date"].dt.to_period("M").astype(str)
df["Year"] = df["Order_Date"].dt.year
df["Profit_Margin"] = df["Profit"] / df["Sales"] * 100
```

### 6. Descriptive Statistics
```python
df[["Quantity","Sales","Profit"]].describe()
```

### 7. Correlation Analysis
```python
df[["Quantity","Sales","Profit"]].corr()
```

### 8. Visualizations
The following charts are included in the `Visualizations` folder:
- Sales by Region
- Monthly Sales Trend
- Profit by Category
- Sales by Category
- Quantity Sold by Product

### 9. Key Metrics
```python
print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Total Quantity:", df["Quantity"].sum())
print("Total Orders:", df["Order_ID"].nunique())
print("Average Order Value:", df["Sales"].sum()/df["Order_ID"].nunique())
```

### 10. Conclusion
The analysis identifies sales trends, profitable categories, regional performance, and product demand. The findings can be used to support inventory planning, regional sales strategies, and customer-focused decisions.
